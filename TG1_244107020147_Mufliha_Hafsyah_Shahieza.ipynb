Perkembangan AI yang begitu pesat, khususnya dalam pelatihan dan pengoperasian model bahasa besar (large language model), ternyata membawa konsekuensi besar terhadap kebutuhan energi global. Berdasarkan laporan International Energy Agency (IEA), konsumsi listrik pusat data (data center) secara global diperkirakan meningkat dari sekitar 415 terawatt jam (TWh) pada tahun 2024 menjadi sekitar 945 TWh pada tahun 2030. Angka tersebut setara dengan total konsumsi listrik tahunan negara Jepang, dengan laju pertumbuhan tahunan mencapai sekitar 15 persen, jauh melampaui rata-rata pertumbuhan konsumsi listrik sektor lain yang hanya berkisar 3 hingga 4 persen. Pertumbuhan yang sangat cepat ini sebagian besar didorong oleh meningkatnya penggunaan server AI, yang konsumsi energinya diperkirakan tumbuh hingga 30 persen setiap tahun. Kondisi ini menegaskan bahwa isu energi AI bukan lagi sekadar persoalan teknis di laboratorium komputer, melainkan telah menjadi isu infrastruktur nasional bahkan geopolitik energi dunia.<br><br>

Riset tersebut menyebutkan bahwa penilaian dampak lingkungan AI selama ini cenderung kurang akurat karena terlalu berfokus pada emisi karbon, sementara jejak penggunaan air dan lahan belum diperhitungkan secara memadai. Padahal, konsumsi air yang berkaitan dengan operasional AI diperkirakan dapat menyamai kebutuhan domestik tahunan dasar dari 1,3 miliar penduduk dunia pada akhir dekade ini. Di samping itu, pertumbuhan AI juga turut menyumbang limbah elektronik (e-waste) serta mendorong aktivitas penambangan mineral langka yang diperlukan untuk memproduksi perangkat keras pendukung AI. Dengan demikian, dampak lingkungan dari AI sesungguhnya bersifat multidimensi dan tidak dapat direduksi hanya pada aspek emisi karbon semata.<br><br>
Untuk mengatasi berbagai dampak tersebut, sejumlah langkah dapat dilakukan baik oleh penyedia layanan AI maupun oleh penggunanya. Terakhir, memperpanjang umur pakai perangkat keras serta mendaur ulang komponen elektronik juga menjadi langkah penting untuk menekan volume limbah elektronik yang dihasilkan industri AI, sehingga pertumbuhan teknologi ini dapat berjalan lebih selaras dengan prinsip keberlanjutan lingkungan.<br><hr>
