## **Langkah 1 - Import Library dan Load Dataset MNIST**

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Input
from tensorflow.keras.utils import to_categorical
import matplotlib.pyplot as plt
import time

# 1. Load dataset MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalisasi data (0-255 → 0-1)
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-hot encoding label
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

print("Shape X_train:", X_train.shape)
print("Shape y_train:", y_train.shape)
```

## **Langkah 2 - Membangun Model JST (Baseline)**

```python
# Bangun model JST (Baseline: 128 - 64)
model = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])

# 3. Kompilasi model
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

## **Langkah 3 - Latih Model dan Evaluasi**

```python
# 4. Latih model
start_time = time.time()

history = model.fit(
    X_train, y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    verbose=1
)

waktu_baseline = time.time() - start_time

# 5. Evaluasi model
loss, acc = model.evaluate(X_test, y_test, verbose=0)
print(f"\nAkurasi pada data uji (Baseline): {acc:.4f}")
print(f"Waktu training: {waktu_baseline:.2f} detik")
```

# **EKSPERIMEN 1 - Mengubah Jumlah Neuron (256 dan 128)**

```python
# Bangun model (Eksperimen 1: 256 - 128)
model_exp1 = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    Dense(256, activation='relu'),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model_exp1.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Latih
start_time = time.time()
history_exp1 = model_exp1.fit(
    X_train, y_train, epochs=10, batch_size=128, validation_split=0.1, verbose=0
)
waktu_exp1 = time.time() - start_time

# Evaluasi
loss_exp1, acc_exp1 = model_exp1.evaluate(X_test, y_test, verbose=0)
print(f"Akurasi (Eksperimen 1: 256-128): {acc_exp1:.4f}")
print(f"Waktu training: {waktu_exp1:.2f} detik")
```

---
# **EKSPERIMEN 2 - Menambah Satu Hidden Layer Lagi**

```python
# Bangun model (Eksperimen 2: 256 - 128 - 64)
model_exp2 = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    Dense(256, activation='relu'),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])

model_exp2.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Latih
start_time = time.time()
history_exp2 = model_exp2.fit(
    X_train, y_train, epochs=10, batch_size=128, validation_split=0.1, verbose=0
)
waktu_exp2 = time.time() - start_time

# Evaluasi
loss_exp2, acc_exp2 = model_exp2.evaluate(X_test, y_test, verbose=0)
print(f"Akurasi (Eksperimen 2: 256-128-64): {acc_exp2:.4f}")
print(f"Waktu training: {waktu_exp2:.2f} detik")
```

---
# **EKSPERIMEN 3 - Perbandingan Sigmoid vs ReLU**

```python
# Fungsi untuk melatih model dengan aktivasi tertentu
def train_with_activation(act):
    m = Sequential([
        Input(shape=(28, 28)),
        Flatten(),
        Dense(128, activation=act),
        Dense(64, activation=act),
        Dense(10, activation='softmax')
    ])
    m.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

    start_time = time.time()
    m.fit(X_train, y_train, epochs=10, batch_size=128, validation_split=0.1, verbose=0)
    waktu = time.time() - start_time

    _, acc = m.evaluate(X_test, y_test, verbose=0)
    return acc, waktu

print("Mulai eksperimen Sigmoid...")
acc_sig, time_sig = train_with_activation('sigmoid')
print(f"Sigmoid -> Akurasi: {acc_sig:.4f}, Waktu: {time_sig:.2f} detik")

print("\nMulai eksperimen ReLU...")
acc_relu, time_relu = train_with_activation('relu')
print(f"ReLU    -> Akurasi: {acc_relu:.4f}, Waktu: {time_relu:.2f} detik")
```

**Kesimpulan Tugas Akhir:**

Berdasarkan seluruh eksperimen yang dilakukan pada dataset MNIST, diperoleh beberapa kesimpulan utama:

1. **Pengaruh Jumlah Neuron (Eksperimen 1):** Menambah jumlah neuron dari baseline (128-64) menjadi (256-128) tidak menjamin kenaikan akurasi. Pada eksperimen ini, akurasi justru sedikit menurun (0.9794) dan waktu training meningkat cukup signifikan (41.34 detik). Hal ini membuktikan bahwa arsitektur baseline sudah cukup optimal untuk dataset MNIST.
2. **Pengaruh Kedalaman Jaringan (Eksperimen 2):** Menambahkan satu hidden layer lagi (256-128-64) justru menurunkan akurasi menjadi 0.9749. Jaringan yang terlalu dalam tanpa regularisasi (seperti Dropout) pada dataset gambar sederhana seperti MNIST cenderung mengalami *overfitting* atau kesulitan konvergensi, serta memakan waktu training paling lama (43.43 detik).
3. **Perbandingan Sigmoid vs ReLU (Eksperimen 3):** Fungsi aktivasi ReLU terbukti lebih unggul secara komprehensif. ReLU menghasilkan akurasi yang lebih tinggi (0.9800) dibandingkan Sigmoid (0.9719). Selain itu, meskipun waktu training keduanya hampir mirip pada dataset ini, Sigmoid secara teori lebih rentan terhadap *vanishing gradient* sehingga kurang efektif untuk jaringan yang lebih dalam.

**Kesimpulan Akhir:**
Model **Baseline (128-64 dengan ReLU)** adalah pilihan yang paling efisien dan optimal untuk klasifikasi MNIST. Model ini menghasilkan akurasi tertinggi (0.9807) dengan waktu training yang relatif cepat (29.26 detik). Penambahan kompleksitas model (neuron atau layer) tidak selalu berbanding lurus dengan peningkatan akurasi, melainkan hanya menambah beban komputasi.
