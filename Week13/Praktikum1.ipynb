## **Langkah 1 - Import Library**

```python
import numpy as np
```

## **Langkah 2 - Menyiapkan Dataset XOR**

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]])

print("Data input (X):")
print(X)
print("\nTarget (y):")
print(y)
```

## **Langkah 3 - Inisialisasi Parameter dan Bobot**

```python
# Parameter jaringan
input_size = 2
hidden_size = 2
output_size = 1
lr = 0.1

# Seed agar hasil reproducible
np.random.seed(42)

# Inisialisasi bobot dan bias
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

print("W1:\n", W1)
print("b1:\n", b1)
print("W2:\n", W2)
print("b2:\n", b2)
```

## **Langkah 4 - Mendefinisikan Fungsi Aktivasi Sigmoid**

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)
```

## **Langkah 5 - Forward Pass dan Backpropagation (Training)**

```python
# Training
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss:.6f}")
```

## **Langkah 6 - Melihat Hasil Prediksi Akhir**

```python
print("Prediksi akhir:")
print(np.round(a2, 4))

print("\nTarget:")
print(y)

loss_final = np.mean(np.square(y - a2))
print(f"\nFinal Loss: {loss_final:.6f}")
```

# **TUGAS 1 - Mengubah Jumlah Neuron Hidden Layer menjadi 3**
## **Langkah 1 - Modifikasi hidden_size menjadi 3**

```python
# Reset seed dan parameter
np.random.seed(42)

input_size = 2
hidden_size_t3 = 3      # diubah menjadi 3
output_size = 1
lr = 0.1

W1_t3 = np.random.randn(input_size, hidden_size_t3)
b1_t3 = np.zeros((1, hidden_size_t3))
W2_t3 = np.random.randn(hidden_size_t3, output_size)
b2_t3 = np.zeros((1, output_size))
```

## **Langkah 2 - Training dengan 3 Neuron Hidden**

```python
for epoch in range(10000):
    z1 = np.dot(X, W1_t3) + b1_t3
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2_t3) + b2_t3
    a2 = sigmoid(z2)

    error = y - a2

    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2_t3.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    W1_t3 += lr * d_W1
    b1_t3 += lr * d_b1
    W2_t3 += lr * d_W2
    b2_t3 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss:.6f}")
```

## **Langkah 3 - Perbandingan Loss Akhir**

```python
pred_t3 = a2
loss_final_t3 = np.mean(np.square(y - pred_t3))

print("Prediksi dengan 3 neuron hidden:")
print(np.round(pred_t3, 4))

print(f"\nFinal Loss (hidden=2): {loss_final:.6f}")
print(f"Final Loss (hidden=3): {loss_final_t3:.6f}")
```

**Analisis:**

| Konfigurasi | Hidden Size | Final Loss | Prediksi Akhir |
|-------------|-------------|------------|----------------|
| Awal        | 2           |  0.128227  | mendekati [0,1,1,0] |
| Tugas 1     | 3           | 0.002297   | mendekati [0,1,1,0] |

Dengan 3 neuron di hidden layer, model memiliki kapasitas representasi lebih besar. Untuk permasalahan XOR yang sederhana, kedua konfigurasi dapat konvergen. Namun jika loss lebih kecil pada hidden=3, ini menunjukkan bahwa penambahan neuron membantu model menemukan representasi fitur yang lebih baik. Jika hasilnya mirip, artinya XOR cukup diselesaikan dengan 2 neuron hidden karena permasalahan hanya membutuhkan satu decision boundary non-linear.

# **TUGAS 2 - Menambahkan Fungsi Aktivasi ReLU**
## **Langkah 1 - Mendefinisikan Fungsi Aktivasi ReLU**

```python
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    # Turunan ReLU dihitung dari nilai pre-activation (z), bukan output (a)
    return (x > 0).astype(float)
```

## **Langkah 2 - Training dengan Hidden Layer ReLU dan Output Sigmoid**

```python
np.random.seed(42)

W1_relu = np.random.randn(input_size, hidden_size) * np.sqrt(2.0 / input_size)
b1_relu = np.zeros((1, hidden_size))
W2_relu = np.random.randn(hidden_size, output_size)
b2_relu = np.zeros((1, output_size))

for epoch in range(10000):
    # Forward pass (hidden pakai ReLU, output pakai Sigmoid)
    z1 = np.dot(X, W1_relu) + b1_relu
    a1 = relu(z1)
    z2 = np.dot(a1, W2_relu) + b2_relu
    a2 = sigmoid(z2)

    error = y - a2

    # Backprop
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2_relu.T) * relu_derivative(z1)  # pakai z1
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    W1_relu += lr * d_W1
    b1_relu += lr * d_b1
    W2_relu += lr * d_W2
    b2_relu += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss:.6f}")
```

## **Langkah 3 - Perbandingan Sigmoid vs ReLU**

```python
pred_relu = a2
loss_final_relu = np.mean(np.square(y - pred_relu))

print("Prediksi dengan ReLU:")
print(np.round(pred_relu, 4))

print(f"\nFinal Loss (Sigmoid): {loss_final:.6f}")
print(f"Final Loss (ReLU):    {loss_final_relu:.6f}")
```

**Analisis:**

| Aktivasi Hidden | Final Loss | Catatan |
|-----------------|------------|---------|
| Sigmoid         | 0.128227   | Konvergen stabil, prediksi mendekati [0, 1, 1, 0] |
| ReLU            |   0.125047 | Konvergensi lebih baik (loss lebih kecil), namun ada satu prediksi yang tertahan di 0.5001 |

Kesimpulan: Berdasarkan eksperimen, ReLU menghasilkan final loss yang lebih kecil (0.125047) dibandingkan Sigmoid (0.128227). Hal ini menunjukkan bahwa ReLU mampu melakukan konvergensi lebih cepat dan efektif pada hidden layer, karena fungsi ini tidak mengalami vanishing gradient seperti yang sering terjadi pada Sigmoid saat nilai input terlalu besar atau terlalu kecil. Meskipun demikian, Sigmoid menghasilkan prediksi yang lebih merata (semua nilai sangat dekat dengan target 0 dan 1), sedangkan ReLU masih menyisakan satu nilai prediksi yang agak jauh (0.5001). Secara keseluruhan, untuk dataset XOR sederhana ini, ReLU terbukti memberikan performa loss yang sedikit lebih unggul.
