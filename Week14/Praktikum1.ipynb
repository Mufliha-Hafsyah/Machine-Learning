```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import numpy as np
from keras.preprocessing import image
import os
```

## **Langkah 2 - Pra Pengolahan Data**

```python
base_dir = '/content/drive/MyDrive/JOBSHEET ML/dataset'

train_dir = os.path.join(base_dir, 'training_set')
test_dir = os.path.join(base_dir, 'test_set')
single_dir = os.path.join(base_dir, 'single_prediction')

print("Isi folder training:", os.listdir(train_dir))
```

### **Langkah 2.1 - Pra Pengolahan Data Training**

```python
train_datagen = ImageDataGenerator(rescale = 1./255,
                                   shear_range = 0.2,
                                   zoom_range = 0.2,
                                   horizontal_flip = True)

training_set = train_datagen.flow_from_directory(train_dir,
                                                 target_size = (64, 64),
                                                 batch_size = 32,
                                                 class_mode = 'binary')
```

### **Langkah 2.2 - Pra Pengolahan Data Testing**

```python
test_datagen = ImageDataGenerator(rescale = 1./255)

test_set = test_datagen.flow_from_directory(test_dir,
                                            target_size = (64, 64),
                                            batch_size = 32,
                                            class_mode = 'binary')
```

## **Langkah 3 - Pembuatan Model CNN**

```python
cnn = tf.keras.models.Sequential()
cnn.add(tf.keras.Input(shape=(64, 64, 3)))

# Layer Konvolusi 1
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu'))
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))

# Layer Konvolusi 2
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu'))
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))

# Flattening
cnn.add(tf.keras.layers.Flatten())

# Fully Connected Layer
cnn.add(tf.keras.layers.Dense(units=128, activation='relu'))
cnn.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))

# Compile
cnn.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
cnn.summary()
```

## **Langkah 4 - Fit CNN**

```python
# Proses training
history = cnn.fit(x = training_set, validation_data = test_set, epochs = 10)
```

## **Langkah 5 - Prediksi dengan 1 Citra**

```python
img_path = os.path.join(single_dir, 'cat_or_dog_2.jpg')

test_image = image.load_img(img_path, target_size = (64, 64))
test_image = image.img_to_array(test_image)
test_image = np.expand_dims(test_image, axis = 0)

result = cnn.predict(test_image)
print("Indeks Kelas:", training_set.class_indices)

if result[0][0] == 1:
  prediction = 'dog'
else:
  prediction = 'cat'

print(f"Hasil Prediksi: {prediction}")
```

**Analisis Praktikum 1:**

Model CNN sederhana berhasil melakukan klasifikasi biner (Cat vs Dog) dengan baik, mencapai akurasi validasi sebesar 77.95%. Penggunaan `ImageDataGenerator` untuk augmentasi (shear, zoom, horizontal flip) terbukti sangat efektif dalam membantu model agar tidak *overfitting* (menghafal data training), yang ditandai dengan selisih yang tipis antara akurasi training (80.09%) dan akurasi validasi (77.95%).

Pada Langkah 5, model berhasil memprediksi gambar `cat_or_dog_2.jpg` sebagai **'cat'** dengan tepat, yang membuktikan bahwa model telah mampu mengekstraksi fitur visual (seperti bentuk telinga, bulu, dan mata) untuk membedakan kedua kelas tersebut.
